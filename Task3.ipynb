### 1)

For en generell Runge-Kutta metode har vi: 

Gitt $b_j$, $c_j$, og $a_{jl}$ for $j,l = 1,\ldots s$

\begin{aligned}
k_{j} &:=
f(t_k + c_j \tau, y_i +  \tau_k \sum_{l=1}^{s} {a}_{jl} k_l)
\quad
j = 1,\ldots s,
\\
y_{k+1} &:= y_{k} + \tau_k \sum_{j=1}^s b_j k_j
\end{aligned}

$\theta$-metoden gir
\begin{aligned}
U^{n+1} &:= U^n + \tau \left( \theta F(t_{n+1}, U^{n+1}) + (1-\theta) F(t_n, U^n) \right)
\quad \text{for } n = 0,1,2,\ldots,N-1.
\end{aligned}
<br>

Vi ser at $\sum_{j=1}^s b_j k_j$ svarer til $\left( \theta F(t_{n+1}, U^{n+1}) + (1-\theta) F(t_n, U^n) \right)$ og identifiserer dermed <br>

\begin{aligned}
b_1 = (1-\theta), \quad b_2 = \theta, \quad k_1 = F(t_n, U^n), \quad k_2 = F(t_{n+1}, U^{n+1}) 
\end{aligned}

For $k_1$ får vi da <br>
\begin{aligned}
k_1 &= F(t_n, U^n)
\\
&= F(t_n + c_1 \tau, U^n +  \tau_k ({a}_{11} k_1 + {a}_{12} k_2))
\end{aligned}

og vi ser at
\begin{aligned}
c_1 = 0, \quad {a}_{11} = 0, \quad {a}_{12} = 0
\end{aligned}

For $k_2$ får vi <br>
\begin{aligned}
k_2 &= F(t_{n+1}, U^{n+1}) 
\\
&= F(t_n + c_2 \tau, U^n +  \tau_k ((1-\theta) F(t_n, U^n) + \theta F(t_{n+1}, U^{n+1}))
\\
&= F(t_n + c_2 \tau, U^n +  \tau_k ({a}_{21} k_1 + {a}_{22} k_2))
\end{aligned}

og vi ser at
\begin{aligned}
c_2 = 1, \quad {a}_{21} = (1-\theta), \quad {a}_{22} = \theta
\end{aligned}

Dette gir følgende Butcher-tableau:

$$
\begin{array}{c|ccc}
0 & 0 & 0
\\  
1 & (1 - \theta) & \theta
\\ 
\hline
& (1 - \theta) & \theta
\end{array}
$$



Dersom $F$ har kontinuerlige deriverte opp til orden $p$, vil Runge-Kutta metoden ha konsistensorden $p$ så lenge betingelsene til og med $p$ er oppfylt:

$$
\begin{array}{|c|c|c}
   \hline
   p  & \text{conditions}  \\ \hline 
   1 & \sum_{i=1}^s b_i = 1 \\ \hline 
  2 & \sum_{i=1}^s b_i c_i = 1/2 \\ \hline 
  3 & \sum_{i=1}^s b_i c_i^2 = 1/3\\ 
   & \sum_{i,j=1}^s b_i a_{ij} c_j = 1/6 
  \\ \hline 
  4 & \sum_{i=1}^s b_ic_i^3=1/4 \\ 
  & \sum_{i,j=1}^s b_i c_i a_{ij}c_j=1/8 \\ 
  & \sum_{i,j=1}^s b_i a_{ij}c_j^2=1/12 \\ 
  & \sum_{i,j,k=1}^s b_i a_{ij} a_{jk} c_k = 1/24 \\ \hline 
\end{array}
$$

Fra Butcher-tableau over ser vi at metoden har konsistensorden $p = 1$, da den første betingelsen er oppfylt for alle $\theta$: <br>
$$
\begin{aligned}
\sum_{i=1}^s b_i = 0 + 1 = 1
\end{aligned}
$$

Den andre betingelsen er derimot ikke oppfylt for alle $\theta$, men kun for $\theta = 1/2$: <br>
\begin{aligned}
\sum_{i=1}^s b_i c_i = (1 - \theta)\cdot 0 + \theta\cdot 1 = \theta = 1/2
\end{aligned}
<br>

Vi ser altså at $\theta$-metoden har konsistensorden $p = 2$, kun for $\theta = 1/2$

Den tredje betingelsen oppfylles ikke, da <br>

\begin{aligned}
\sum_{i=1}^s b_i c_i^2 = (1 - \theta)\cdot 0^2 + \theta\cdot 1^2 = \theta = 1/2 \neq 1/3
\end{aligned}

For $\theta = 0$ får vi 
\begin{aligned}
U^{n+1} &= U^n + \tau F(t_n, U^n)
\quad \text{for } n = 0,1,2,\ldots,N-1.
\end{aligned}
<br>
som vi kjenner igjen som eksplisitt Eulers metode. <br>

For $\theta = 1$ får vi 
\begin{aligned}
U^{n+1} &:= U^n + \tau F(t_{n+1}, U^{n+1})
\quad \text{for } n = 0,1,2,\ldots,N-1.
\end{aligned} 
<br>
som vi kjenner igjen som implisitt Eulers metode.

Både eksplisitt og implisitt Eulers metode har kjent konsistensorden 1, noe som stemmer godt overens med resultatene ovenfor.

For $\theta = 1/2$ får vi 
\begin{aligned}
U^{n+1} &:= U^n + \tau \left( \frac{1}{2}F(t_n, U^n) + \frac{1}{2}F(t_{n+1}, U^{n+1}) \right) 
&= U^n + \frac{\tau}{2} \left(F(t_n, U^n) + F(t_{n+1}, U^{n+1}) \right)
\quad \text{for } n = 0,1,2,\ldots,N-1.
\end{aligned}
<br>
som vi kjenner igjen som trapes-metoden. Denne har konsistensorden 2, som også stemmer overens med resultatene ovenfor. <br>

### 2)

Vi ønsker å bestemme stablilitetsfunksjonen $r_{\theta}(z)$ til $\theta$-metoden for å analysere hvilke verdier som gir en stabil numerisk metode. 
Til dette bruker vi en test-ODE, 
$$
\frac{\partial U}{\partial t} = \lambda U = F(t_n, U^n),  \quad U(0) = U^0
$$
som har kjent løsning 
$$ 
U(t) = U^0 e^{\lambda t}
$$
<br>

Ved å sette dette inn i $\theta$ -metoden får vi <br>

$$
U^{n+1} = U^n + \tau \left( \theta \lambda U^{n+1} + (1-\theta) U^n \right)
= U^n + \tau \theta \lambda U^{n+1} + \tau (1 - \theta) \lambda U^n
$$
$$
\implies (1 - \tau \theta \lambda) U^{n+1} = (1 + \tau (1 - \theta) \lambda) U^n
$$
$$
\implies U^{n+1} = \frac{(1 +  (1 - \theta)\tau \lambda)}{(1 -  \theta \tau \lambda)} U^n
$$

Vi har dermed stabilitetsfunksjonen
$$
r_{\theta}(z) = \frac{1 + (1 - \theta) z}{1 - \theta z} , \quad z = \tau \lambda
$$

For å finne stabilitetsområdet til $\theta$-metoden må vi se på de z-verdiene som gir $\lvert r_{\theta}(z) \rvert \le 1$

```python
import numpy as np 
import matplotlib.pyplot as plt

xlim = (-5, 5)
ylim = (-5, 5)
resolution = 500

theta_array = [0, 0.25, 0.498, 0.5, 0.502, 0.75, 1]

def r_theta(z):                             # Definerer stabilitetsfunksjonen
    r = (1 + (1-theta)*z)/(1-theta*z)
    return r

# Lager grid av komplekse tall
x = np.linspace(xlim[0], xlim[1], resolution)
y = np.linspace(ylim[0], ylim[1], resolution)
X, Y = np.meshgrid(x, y)
Z = X + 1j * Y
    
# Plotter stabilitetsområdet for hver teta-verdi

fig, axs = plt.subplots(2, 4, figsize=(15,8))
axs = axs.flatten()

for i in range(len(theta_array)):
    theta = theta_array[i]
    R = np.abs(r_theta(Z))                        # Beregner |r(z)| på grid

    axs[i].contourf(X, Y, R, levels=[0, 1], colors=['mediumseagreen'], alpha=1)
    axs[i].contour(X, Y, R, levels=[1], colors=['black'])

    axs[i].axhline(0, color='black', linewidth=0.5)
    axs[i].axvline(0, color='black', linewidth=0.5)
    axs[i].set_xlim(xlim)
    axs[i].set_ylim(ylim)
    axs[i].set_title(f'θ = {theta}')
    axs[i].grid()

fig.delaxes(axs[-1])
fig.suptitle("Stabilitetsområde for θ-metoden for ulike θ", fontsize=18)
fig.supxlabel("Re(z)")
fig.supylabel("Im(z)")

plt.tight_layout()
plt.show()
```

Ovenfor ser vi et plot av stabilitetsområdet til $\theta$-metoden for ulike $\theta$-verdier. 

En A-stabil metode defineres ved at $\lvert r_{\theta}(z) \rvert \le 1$ for alle $Re(z) \le 0$, altså at stabilitetsområdet inkluderer hele det komplekse halvplanet med negativ realdel. <br> 
Dette betyr at metoden er stabil for alle negative reelle $\lambda$, uavhengig av størrelsen på tidssteget $\tau$. 

Vi ser at $\theta$-metoden ser ut til å være A-stabil for $\theta = 0.498$, $\theta = 0.5$, $\theta = 0.502$, $\theta = 0.75$ og $\theta = 1$.
Generelt kan vi dermed anta at $\theta$-metoden må være stabil for $\theta \gtrsim 0.5$. For $\theta \lesssim 0.5$ ser vi at metoden får mindre og mindre stabilitetsområde. Dette betyr at metoden kun er stabil for lave $z = \lambda \tau$, og man må typisk bruke svært små tidssteg for å unngå eksponensiell vekst i feilen.

### 3)

For å bestemme stabilitetsområdet til $\theta$-metoden matematisk finner vi sirkelen som beskriver grensen mellom det stabile og ustabile området. 
Denne grensen er gitt av $$\partial S_{\theta} = \{ z \in \mathbb{C} : \lvert r_{\theta}(z) \rvert = 1 \}$$

\begin{aligned}
\lvert r_{\theta}(z) \rvert &= 1
\\
\frac{\lvert 1 + (1 - \theta) z \rvert}{\lvert 1 - \theta z \rvert} &= 1
\\
\lvert 1 + (1 - \theta) z \rvert &= \lvert 1 - \theta z \rvert
\end{aligned}

Kvadrerer begge sider ved $\lvert z \rvert^2 = z \cdot \bar{z}$:

\begin{aligned}
(1 + (1 - \theta)z)(1 + (1 - \theta) \bar{z}) &= (1 - \theta z)(1 - \theta \bar{z})
\\
1 + (1 - \theta) \bar{z} + (1 - \theta)z + (1 - \theta)^2 \lvert z \rvert^2 &= 1 - \theta \bar{z} - \theta z + \theta^2 \lvert z \rvert^2
\\
1 + \bar{z} - \theta \bar{z} + z - \theta z + \lvert z \rvert^2 - 2 \theta \lvert z \rvert^2 + \theta^2 \lvert z \rvert^2 &= 
1 - \theta \bar{z} - \theta z + \theta^2 \lvert z \rvert^2
\\
\lvert z \rvert^2 - 2 \theta \lvert z \rvert^2 + z + \bar{z} &= 0 
\end{aligned}

og bruker at $\lvert z \rvert^2 = x^2 + y^2$ og $z + \bar{z} = x + iy + x - iy = 2x$:

\begin{aligned}
x^2 + y^2 - 2 \theta (x^2 + y^2) + 2x &= 0
\\
(x^2 + y^2)(1 - 2 \theta) + 2x &= 0
\\
x^2 + y^2 + \frac{2x}{1 - 2 \theta} &= 0
\\
x^2 + 2x \frac{1}{1 - 2 \theta} + (\frac{1}{1 - 2 \theta})^2 + y^2 &= (\frac{1}{1 - 2 \theta})^2
\\
(x^2 + \frac{1}{1 - 2 \theta})^2 + y^2 = (\frac{1}{1 - 2 \theta})^2 
\end{aligned}

Vi ser at denne likningen representerer en sirkel i det komplekse planet med 
\begin{aligned}
R = \frac{1}{\lvert 1 - 2 \theta \rvert}
\end {aligned}

og sentrum i 
\begin{aligned}
(- \frac{1}{1 - 2 \theta}, \quad 0)
\end{aligned}

Sirkelen beskriver randen til stabilitetsområdet til $\theta$-metoden, som vi også kan se i plottet over. 

Vi ser at uttrykkene stemmer godt overens med plottene da for eksempel $\theta = 0$ gir en sirkel med $R = 1$ og sentrum i $C = (-1, 0)$, og økende $\theta$ vil gi større og større sirkel fram til $\theta = 0.5$.

For $\theta \to 0.5^-$ får vi $R \to \infty$ og $C \to (- \infty, 0)$, mens for $\theta \to 0.5^+$ får vi $R \to \infty$ og $C \to (\infty, 0)$, noe som også stemmer godt overens med plottene over. 

For $\theta \geq 0.5$ ser vi at sentrum flytter seg til det positive halvplanet. For for eksempel $\theta = 0.75$ får vi en sirkel med $R = \frac{1}{\lvert 1 - 2 \cdot 0.75 \rvert} = 2$ og sentrum i $C = (2, 0)$.

#### Beregningsmessig oppgave 1

For å få en løselig differensialligning, fourier transformerer vi den tidsavhengige biharmoniske ligning:
$$
\mathcal{F}(\partial_t u + \kappa \Delta^2 u = g) \implies \partial_t \hat{u} = \hat{g} - \kappa \left| k \right|^4 \hat{u}
= F(t, \hat{u}) = F(\hat{u})
$$

Så bruker vi $\theta$ - metoden for å løse ligningen ved å sette inn $\partial_t \hat{u} = \hat{g} - \kappa \left| k \right|^4 \hat{u}$, inn for $F(t_n, U^n)$, og får etter litt regning:
$$
\hat{u}^{n+1} = \frac{(1 - \tau (1 - \theta) \kappa \left| k \right|^4) \hat{u}^n + \tau (\theta \hat{g}^{n+1} - (1 - \theta) \hat{g}^{n})}{1 + \tau \theta \kappa \left| k \right|^4}
$$  
Med dette uttrykket kan man da løse $\hat{u}$ stegvis.

## Importerer biblioteker, implementerer løser og definerer globale variabler

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from scipy.fft import fft, fft2, ifft, ifft2, fftfreq, fftshift
import pandas as pd
```

```python
def transient_biharmonic_solver(*, kappa, X, Y, U0, t0, T, Nt, Lx, Ly, theta=0.5, g=None):

    # Lager frekvensrommet
    Nx, Ny = X.shape 
    kx = fftfreq(Nx, d=Lx/Nx)*2*np.pi
    ky = fftfreq(Ny, d=Ly/Ny)*2*np.pi
    KX, KY = np.meshgrid(kx, ky, sparse=True)

    # Definerer k**4 til senere bruk
    k4 = (KX**2 + KY**2)**2

    # Finner initialbetingelsene i frekvensrommet og tidssteg
    U_hat0 = fft2(U0)
    U_hat = U_hat0
    t = t0
    dt = (T-t0)/Nt

    yield U_hat0, t     # Yielder init. bet. i frekvensrom
    
    if g is None:       # Håndterer tilfellet at g ikke er en kallbar funksjon (Homogen likning)
        g = lambda x, y, t: np.zeros_like(U_hat)
    
    while t < T-dt/2:
        g_now = fft2(g(X, Y, t))        # g(n)
        g_next = fft2(g(X , Y, t+dt))   # g(n+1)
        
        # Benytter løsning av lineær likning for å finne U(n+1) implisitt
        U_hat = (U_hat+dt*(theta*g_next+(1-theta)*(g_now - kappa*k4 * U_hat)))/(1+dt*theta*kappa*k4)
        
        t+=dt # Oppdaterer tid

        yield U_hat, t # Yielder nåværende løsning

        
        
        
```

```python
Nx, Ny = 20, 20             # Antall romlige sampling punkter
Lx, Ly = 2*np.pi, 2*np.pi   # Størrelse på romlig domene
lam = 4                     # Den beregnede lambda som gir homogent tilfelle g=0
kappa = 1 
t0 = 0
T = 1

# Definerer romlig domene symmetrisk om origo
x = np.linspace(-Lx/2, Lx/2, Nx, endpoint=False) 
y = np.linspace(-Ly/2, Ly/2, Ny, endpoint=False)
X, Y = np.meshgrid(x, y, indexing='ij')  # Shape: (Nx, Ny)


Nts = [10, 20, 40, 80, 160, 320, 640]
```

## Funksjon for L-inf-L-inf feil & eksperimentell konvergens orden og plotting av resultater i tabell

```python
def error(Nts, thetas, u_ex, g=None): # Takes lists of Nt values and theta values, and callable u_ex and g
        # Gjør om lister til array for å kunne bruke dem i utregning av eoc
        Nts = np.array(Nts) 
        theta = np.array(thetas)
        
        # Lager arrays som skal holde resultatene
        errors = np.zeros((len(thetas), len(Nts)))
        eocs = np.zeros((len(thetas), len(Nts)-1))

        U0 = u_ex(X, Y, t0) # Initialbetingelse for løseren
        for i in range(len(thetas)):
                timeAndSpaceMax = np.zeros(len(Nts)) # L-inf feil, altså maksimal feil over alle tidssteg og over hele rommet
                for j in range(len(Nts)):

                        # Nåværende parametere for løseren
                        Nt = Nts[j]
                        theta = thetas[i]
                        solver = transient_biharmonic_solver(kappa=kappa, 
                                                                X=X, Y=Y, U0=U0, 
                                                                t0=t0, T=T, Nt=Nt,
                                                                Lx=Lx, Ly=Ly,
                                                                theta=theta,
                                                                g=g)
                        
                        spacialMax = np.zeros(Nt+1) # Array til maksimale verdier over hele rommet ved hvert tidssteg
                        k = 0
                        for Uhat, t in solver: 
                                
                                U = ifft2(Uhat).real # Finner U i fysisk rom ved invers fourier transform
                                spacialMax[k] = ((np.abs(u_ex(X, Y, t) - U).max())) # Finner størst feil i rommet ved tidssteget
                                k+=1
                        timeAndSpaceMax[j] = (spacialMax).max() # Finner størst feilen som er størst blant feilene for alle tidsstegene
                errors[i] = timeAndSpaceMax # Lagrer listen med feil for nåværende verdi av theta og alle verdier av Nt
        
                eocs[i] = np.log(errors[i][1:]/errors[i][:-1])/np.log(Nts[:-1]/Nts[1:]) # Beregner EOC basert på den funnede feilen

        return  errors, eocs # Returnerer (thetas x Nts) formet array med L-inf feil og EOCs respoektivt for hver theta og Nt

def table(Nts, thetas, u_ex, g = None): # Tar inn liste emd Nt og theta verdier og kallbare u_ex og g funksjoner
        
        errors, eocs = error(Nts, thetas, u_ex, g=g) # Finner feil og konvergens orden

        for i in range(len(errors)): # Lager tabeller for hver theta verdi
                df = pd.DataFrame()
                df['Nt'] = Nts
                df[f'L-inf error, θ = {thetas[i]}'] = errors[i]
                eoc_padded = np.append([np.nan], eocs[i])

                df[f'EOC, θ = {thetas[i]}'] = eoc_padded
                #df = df.style.set_caption(f"theta = {thetas[i]}").set_table_styles(styles)

                # show datafram in a better format
                display(df)
```

## CFL krav

Observer at for å unngå vekst må tau < 2/lamda altså Nt > kappa*k^4/2 = 20 000

CFL krav forteller oss noe om sammenhengen mellom stabilitetsfunksjonen og stegstørrelse for en løsningsmetode.
Vi tar utgangspunkt i Un+1 = (1-tau(1-teta)lambda)/(1+tau* teta* lambda) * Un. Herfra kan vi se at veksten mellom steg kan måles med G, der abs(G) = abs(Un+1/Un)= abs((1-tau(1-teta)lambda)/(1+tau * teta*lambda)). Og dette forholdet burde være mindre enn 1 for å unngå at løsningen plutselig eksploderer slik den gjør i stive problemer. Ser vi på teta = 0 ser vi at vi må ha abs(1-tau * lambda)<=1 og dermed 0<= tau * lamda <= 2. Med lambda lik kappa * k4 slik den er for den biharmoniske operator, gir dette kravet tau <= 2/(kappa * k4). Altså Nt > kappa *k4/2. Vi finner dette ved å begrense med Ncfl = k4max / 2

```python
# Lager frekvensrommet
Nx, Ny = X.shape 
kx = fftfreq(Nx, d=Lx/Nx)*2*np.pi
ky = fftfreq(Ny, d=Ly/Ny)*2*np.pi
KX, KY = np.meshgrid(kx, ky, sparse=True)

# Definerer k**4 til senere bruk
k4 = (KX**2 + KY**2)**2
Ncfl = int(np.max(k4)/2)

theta = 0
Ncfls = [int(0.5*Ncfl), Ncfl, 2*Ncfl, 4*Ncfl]

u_ex = lambda x, y, t: np.sin(x)*np.cos(y)*np.exp(-lam * kappa * t)
theta = [1]         
table(Ncfls, theta, u_ex)
```

```python
# Bruker kode fra forelesningsnotater for å finne g
def manufacture_solution_biharmonic(u_str, kappa, M, f):
    import sympy as sy
    from sympy import sin, cos, exp
    x, y, t = sy.symbols('x y t')
    u_sy = eval(u_str) # symbolsk representasjon av u evalueres fra strengen
    laplace = lambda u: sy.diff(u, x, x) + sy.diff(u, y, y) # Definerer laplace
    g_sy =  sy.diff(u_sy, t) - sy.simplify(M*laplace(f(u_sy)))+sy.simplify(kappa*(laplace(laplace(u_sy))))  # skriver uttrykk for g
    
    # funksjonene gjøres til kallbare funksjoner og printes
    u = sy.lambdify((x, y, t), u_sy, modules='numpy') 
    g = sy.lambdify((x, y, t), g_sy, modules='numpy')
    print(f'u = {u_sy}') 
    print(f'u0 = {u_sy.subs(t, 0)}')
    print(f'g = {g_sy}')
    return u, g

M = 1
f = lambda u : u**3 - u
u_ex_str = '(exp(1+sin(x)*sin(x))+exp(1+cos(y)*cos(y)))*exp(-4*kappa*t)'
u_ex, g = manufacture_solution_biharmonic(u_ex_str, kappa, M, f)
```
