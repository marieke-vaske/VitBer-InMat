### 1)

For en generell Runge-Kutta metode har vi: 

Gitt $b_j$, $c_j$, og $a_{jl}$ for $j,l = 1,\ldots s$

\begin{aligned}
k_{j} &:=
f(t_k + c_j \tau, y_i +  \tau_k \sum_{l=1}^{s} {a}_{jl} k_l)
\quad
j = 1,\ldots s,
\\
y_{k+1} &:= y_{k} + \tau_k \sum_{j=1}^s b_j k_j
\end{aligned}

$\theta$-metoden gir
\begin{aligned}
U^{n+1} &:= U^n + \tau \left( \theta F(t_{n+1}, U^{n+1}) + (1-\theta) F(t_n, U^n) \right)
\quad \text{for } n = 0,1,2,\ldots,N-1.
\end{aligned}
<br>

Vi ser at $\sum_{j=1}^s b_j k_j$ svarer til $\left( \theta F(t_{n+1}, U^{n+1}) + (1-\theta) F(t_n, U^n) \right)$ og identifiserer dermed <br>

\begin{aligned}
b_1 = (1-\theta), \quad b_2 = \theta, \quad k_1 = F(t_n, U^n), \quad k_2 = F(t_{n+1}, U^{n+1}) 
\end{aligned}

For $k_1$ får vi da <br>
\begin{aligned}
k_1 &= F(t_n, U^n)
\\
&= F(t_n + c_1 \tau, U^n +  \tau_k ({a}_{11} k_1 + {a}_{12} k_2))
\end{aligned}

og vi ser at
\begin{aligned}
c_1 = 0, \quad {a}_{11} = 0, \quad {a}_{12} = 0
\end{aligned}

For $k_2$ får vi <br>
\begin{aligned}
k_2 &= F(t_{n+1}, U^{n+1}) 
\\
&= F(t_n + c_2 \tau, U^n +  \tau_k ((1-\theta) F(t_n, U^n) + \theta F(t_{n+1}, U^{n+1}))
\\
&= F(t_n + c_2 \tau, U^n +  \tau_k ({a}_{21} k_1 + {a}_{22} k_2))
\end{aligned}

og vi ser at
\begin{aligned}
c_2 = 1, \quad {a}_{21} = (1-\theta), \quad {a}_{22} = \theta
\end{aligned}

Dette gir følgende Butcher-tableau:

$$
\begin{array}{c|ccc}
0 & 0 & 0
\\  
1 & (1 - \theta) & \theta
\\ 
\hline
& (1 - \theta) & \theta
\end{array}
$$



Dersom $F$ har kontinuerlige deriverte opp til orden $p$, vil Runge-Kutta metoden ha konsistensorden $p$ så lenge betingelsene til og med $p$ er oppfylt:

$$
\begin{array}{|c|c|c}
   \hline
   p  & \text{conditions}  \\ \hline 
   1 & \sum_{i=1}^s b_i = 1 \\ \hline 
  2 & \sum_{i=1}^s b_i c_i = 1/2 \\ \hline 
  3 & \sum_{i=1}^s b_i c_i^2 = 1/3\\ 
   & \sum_{i,j=1}^s b_i a_{ij} c_j = 1/6 
  \\ \hline 
  4 & \sum_{i=1}^s b_ic_i^3=1/4 \\ 
  & \sum_{i,j=1}^s b_i c_i a_{ij}c_j=1/8 \\ 
  & \sum_{i,j=1}^s b_i a_{ij}c_j^2=1/12 \\ 
  & \sum_{i,j,k=1}^s b_i a_{ij} a_{jk} c_k = 1/24 \\ \hline 
\end{array}
$$

Fra Butcher-tableau over ser vi at metoden har konsistensorden $p = 1$, da den første betingelsen er oppfylt for alle $\theta$: <br>
$$
\begin{aligned}
\sum_{i=1}^s b_i = 0 + 1 = 1
\end{aligned}
$$

Den andre betingelsen er derimot ikke oppfylt for alle $\theta$, men kun for $\theta = 1/2$: <br>
\begin{aligned}
\sum_{i=1}^s b_i c_i = (1 - \theta)\cdot 0 + \theta\cdot 1 = \theta = 1/2
\end{aligned}
<br>

Vi ser altså at $\theta$-metoden har konsistensorden $p = 2$, kun for $\theta = 1/2$

Den tredje betingelsen oppfylles ikke, da <br>

\begin{aligned}
\sum_{i=1}^s b_i c_i^2 = (1 - \theta)\cdot 0^2 + \theta\cdot 1^2 = \theta = 1/2 \neq 1/3
\end{aligned}

For $\theta = 0$ får vi 
\begin{aligned}
U^{n+1} &= U^n + \tau F(t_n, U^n)
\quad \text{for } n = 0,1,2,\ldots,N-1.
\end{aligned}
<br>
som vi kjenner igjen som eksplisitt Eulers metode. <br>

For $\theta = 1$ får vi 
\begin{aligned}
U^{n+1} &:= U^n + \tau F(t_{n+1}, U^{n+1})
\quad \text{for } n = 0,1,2,\ldots,N-1.
\end{aligned} 
<br>
som vi kjenner igjen som implisitt Eulers metode.

Både eksplisitt og implisitt Eulers metode har kjent konsistensorden 1, noe som stemmer godt overens med resultatene ovenfor.

For $\theta = 1/2$ får vi 
\begin{aligned}
U^{n+1} &:= U^n + \tau \left( \frac{1}{2}F(t_n, U^n) + \frac{1}{2}F(t_{n+1}, U^{n+1}) \right) 
&= U^n + \frac{\tau}{2} \left(F(t_n, U^n) + F(t_{n+1}, U^{n+1}) \right)
\quad \text{for } n = 0,1,2,\ldots,N-1.
\end{aligned}
<br>
som vi kjenner igjen som trapes-metoden. Denne har konsistensorden 2, som også stemmer overens med resultatene ovenfor. <br>

### 2)

Vi ønsker å bestemme stablilitetsfunksjonen $r_{\theta}(z)$ til $\theta$-metoden for å analysere hvilke verdier som gir en stabil numerisk metode. 
Til dette bruker vi en test-ODE, 
$$
\frac{\partial U}{\partial t} = \lambda U = F(t_n, U^n)
$$

Ved å sette dette inn i $\theta$-metoden får vi <br>

$$
U^{n+1} = U^n + \tau \left( \theta \lambda U^{n+1} + (1-\theta) U^n \right)
= U^n + \tau \theta \lambda U^{n+1} + \tau (1 - \theta) \lambda U^n
= \frac{1 + \tau (1 - \theta) \lambda}{1 - \tau \theta \lambda} U^n
= \frac{1 + (1 - \theta) z}{1 - \theta z} U^n
$$

Vi har dermed stabilitetsfunksjonen
$$
r_{\theta}(z) = \frac{1 + (1 - \theta) z}{1 - \theta z} , \quad z = \tau \lambda
$$

For å finne stabilitetsområdet til $\theta$-metoden må vi se på de z-verdiene som gir $\lvert r_{\theta}(z) \rvert \le 1$

```python
import numpy as np 
import matplotlib.pyplot as plt 

xlim = (-3, 3)
ylim = (-3, 3)
resolution = 500

theta_array = [0, 0.25, 0.498, 0.5, 0.502, 0.75, 1]

def r_theta(z):                             # Definerer stabilitetsfunksjonen
    r = (1 + (1-theta)*z)/(1-theta*z)
    return r

# Lager grid av komplekse tall
x = np.linspace(xlim[0], xlim[1], resolution)
y = np.linspace(ylim[0], ylim[1], resolution)
X, Y = np.meshgrid(x, y)
Z = X + 1j * Y
    
# Plotter stabilitetsområdet for hver teta-verdi

fig, axs = plt.subplots(2, 4, figsize=(15,8))
axs = axs.flatten()

for i in range(len(theta_array)):
    theta = theta_array[i]
    R = np.abs(r_theta(Z))                        # Beregner |r(z)| på grid

    axs[i].contourf(X, Y, R, levels=[0, 1], colors=['mediumseagreen'], alpha=1)
    axs[i].contour(X, Y, R, levels=[1], colors=['black'])

    axs[i].axhline(0, color='black', linewidth=0.5)
    axs[i].axvline(0, color='black', linewidth=0.5)
    axs[i].set_xlim(xlim)
    axs[i].set_ylim(ylim)
    axs[i].set_title(f'θ = {theta}')
    axs[i].grid()

fig.delaxes(axs[-1])

plt.tight_layout()
plt.show()
```
